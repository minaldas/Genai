```python
#Core LLamapare imports
from llama_parse import LlamaParse
from pathlib import Path


parser = LlamaParse(result_type="markdown",verbose=True)
SAMPLE_DIR = Path("./Data")
pdf_path = str(SAMPLE_DIR /"302- Minal Das.pdf")

if Path(pdf_path).exists():
    print("inside if condition")
    print(f"Parsing: {pdf_path}")
    print("-"*50)

    documents = parser.load_data(pdf_path)

    print(f"\n Parsing Complete")
    print(f"Number of documents returned: {len(documents)}")

    if documents:
        doc = documents[0]
        print(f"\n Document type: {type(doc).__name__}")
        print(f"Content length: {len(doc.text)} characters")
        print("="* 50)
        print(doc.text)
else:
    print(f"File not found: {pdf_path}")
        
```
